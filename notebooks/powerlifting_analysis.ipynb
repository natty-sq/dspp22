```python
def remove_iqr_outliers(group):
    Q1 = group["TotalKg"].quantile(0.25)
    Q3 = group["TotalKg"].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return group[(group["TotalKg"] >= lower) & (group["TotalKg"] <= upper)]
```

```python
df_clean = df_sbd.groupby("AgeClass", group_keys=False).apply(remove_iqr_outliers)
```

```python
df_clean = df_clean.groupby("BodyweightKg", group_keys=False).apply(remove_iqr_outliers)
```

```python
df_clean = df_clean.groupby("WeightClassKg", group_keys=False).apply(remove_iqr_outliers)
```

```python
age_order = [
    "18-19", "20-23", "24-34",
    "35-39", "40-44", "45-49",
    "50-54", "55-59", "60-64",
    "65-69", "70-74", "75-79", "80+"
]

df_sbd["AgeClass"] = pd.Categorical(
    df_sbd["AgeClass"],
    categories=age_order,
    ordered=True
)
```

```python
bw_vs_total["BW_to_Total_pct"] = (bw_vs_total["BodyweightKg"] / bw_vs_total["TotalKg"]) * 100
```

```python
ratio_by_sex = bw_vs_total.groupby("Sex")["BW_to_Total_pct"].mean()
print(ratio_by_sex)
```

```python
grouped = (
    bw_vs_total
    .groupby(["WeightClassKg", "Sex"])
    .agg(
        avg_bodyweight=("BodyweightKg", "mean"),
        avg_total=("TotalKg", "mean")
    )
)
```

```python
grouped["ratio_pct"] = (grouped["avg_bodyweight"] / grouped["avg_total"]) * 100
```

```python
grouped
```

```python
grouped = grouped.reset_index()

grouped["WeightClassKg"] = pd.Categorical(
    grouped["WeightClassKg"],
    categories=weight_order,
    ordered=True
)
```

```python
plt.figure(figsize=(14, 6))
sns.barplot(
    data=grouped.reset_index(),
    x="WeightClassKg",
    y="ratio_pct",
    hue="Sex"
)

plt.ylabel("BW / Total Ratio (%)")
plt.title("Average Bodyweight-to-Total Lifted Ratio by Weight Class and Sex")
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.scatterplot(
    data=grouped,
    x="WeightClassKg",
    y="ratio_pct",
    hue="Sex",
    style="Sex",
    s=120
)

plt.title("Bodyweight to total lifted ratio (%) by Weight Class and Sex")
plt.xlabel("Weight Class (kg)")
plt.ylabel("Ratio (%)")
plt.grid(True, alpha=0.3)
plt.show()
```
